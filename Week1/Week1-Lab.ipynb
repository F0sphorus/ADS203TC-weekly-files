```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
# 1. Define parameters
fs = 100 # Sampling frequency in Hz
f0 = 20 # Signal frequency in Hz
A = 1.0 # Amplitude
phi = np.pi / 4 # Phase
duration = 1.0 # seconds

# 2. Generate sample indices
num_samples = int(duration * fs)
n = np.arange(num_samples)

# 3. Generate the discrete-time signal
x_n = A * np.cos(2 * np.pi * f0 * n / fs + phi)

# 4. Plotting
plt.figure(figsize=(12, 6))
```

```python
# Subplot 1: Line plot
plt.subplot(2, 1, 1)
plt.plot(n, x_n)
plt.title('Sinusoidal Signal (Line Plot)')
plt.xlabel('Sample Index (n)')
plt.ylabel('Amplitude')
plt.grid(True)
```

```python
# Subplot 2: Stem plot
plt.subplot(2, 1, 2)
plt.stem(n, x_n)
plt.title('Sinusoidal Signal (Stem Plot)')
plt.xlabel('Sample Index (n)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# 1.Define signal parameters
f0 = 5.0 # Frequency of the sinusoid in Hz
A = 1.5 # Amplitude
duration = 2.0 # Duration in seconds

# 2. Define sampling rates
fs_high = 1000.0 # High sampling rate for "continuous" feel
fs_discrete = 25.0 # Lower sampling rate for the discrete signal

# 3. Generate the "continuous" signal
# Create a fine-grained time vector
t_cont = np.arange(0, duration, 1/fs_high)
x_cont = A * np.sin(2 * np.pi * f0 * t_cont)

# 4. Generate the discrete signal
# Create a coarser time vector for the samples
t_disc = np.arange(0, duration, 1/fs_discrete)
x_disc = A * np.sin(2 * np.pi * f0 * t_disc)

# 5. Plotting
plt.figure(figsize=(12, 6))
```

```python
# Plot the "continuous" signal as a line
plt.plot(t_cont, x_cont, 'b-', label=f'Continuous-Time Signal (Simulated at {fs_high} Hz)')

# Plot the discrete signal using a stem plot
plt.stem(t_disc, x_disc, linefmt='r-', markerfmt='ro', basefmt='r-',
         label=f'Discrete-Time Signal (Sampled at {fs_discrete} Hz)')

plt.title('Comparison of Continuous and Discrete Signal Representations')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Parameters
f0 = 40.0 # Original frequency
fs_under = 50.0 # Undersampling frequency
duration = 0.2

# Generate high-resolution signal for reference
fs_high = 1000.0
t_high = np.arange(0, duration, 1/fs_high)
x_high = np.cos(2 * np.pi * f0 * t_high)

# Generate undersampled signal
t_under = np.arange(0, duration, 1/fs_under)
x_under = np.cos(2 * np.pi * f0 * t_under)

# Plotting to show aliasing
plt.figure(figsize=(12, 6))
```

```python
# Plot 1: Original signal with undersampled points
plt.subplot(2, 1, 1)
plt.plot(t_high, x_high, label=f'Original Signal ({f0} Hz)')
plt.plot(t_under, x_under, 'ro', label=f'Sampled at {fs_under} Hz')
plt.title('Original Signal and Undersampled Points')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
```

```python
# Plot 2: Reconstructed from undersampled points
plt.subplot(2, 1, 2)
plt.plot(t_under, x_under, 'r-o')
# Calculate expected alias frequency: |40-50| = 10 Hz
f_alias = 10.0
plt.title(f'Apparent Signal from Undersampling (Alias at {f_alias} Hz)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Parameters
fs = 200
duration = 1.0
n = np.arange(duration * fs)

# Signal 1
f1, A1 = 10, 1.0
x1_n = A1 * np.sin(2 * np.pi * f1 * n / fs)

# Signal 2
f2, A2 = 25, 0.5
x2_n = A2 * np.sin(2 * np.pi * f2 * n / fs)

# Composite signal
y_n = x1_n + x2_n

# Generate noise
noise_amplitude = 2.0
noise = noise_amplitude * np.random.randn(len(n))

# Add noise to signal
z_n = y_n + noise
```

```python
# Plotting
plt.figure(figsize=(12, 6))
plt.plot(n / fs, y_n, label='Clean Composite Signal')
plt.plot(n / fs, z_n, label='Noisy Signal', alpha=0.7)
plt.title('Composite Signal with Additive White Gaussian Noise')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```
